# NNQS (Neural Network Quantum States) 

NNQS (Neural Network Quantum States) — это класс вариационных квантовых состояний, параметры которых задаются с помощью искусственной нейронной сети. Этот метод используется для компьютерного моделирования сложных многочастичных квантовых систем.

Нейросеть выступает в роли вариационного анзаца (приближения), способного гибко и эффективно описывать многочастичную волновую функцию. Это позволяет решать задачи квантовой механики и квантовой химии, обходя проблему экспоненциального роста сложности вычислений (так называемое «проклятие размерности»).

# Сферы применения метода
Метод активно развивается и применяется в следующих областях:

Квантовая химия: расчет энергий основного состояния молекул и предсказание поверхностей потенциальной энергии с высокой точностью.

Физика конденсированного состояния: симуляция сильно коррелированных твердотельных систем (например, одномерных цепочек водорода, кристаллической решетки алмаза).

Физика высоких энергий: изучение спектроскопии экзотических адронов (многокварковых состояний) и решение релятивистских волновых уравнений вроде уравнения Клейна — Гордона — Фока.

# История 
1. До 2016 года 
Десятилетиями физики искали эффективные способы сжатия описания многочастичных квантовых систем. Матричные произведения (MPS) и тензорные сети (Tensor Networks) отлично работали для одномерных цепочек (1D), но пасовали перед двумерными (2D) и трехмерными (3D) системами со сложным, «объемным» законом запутанности (Volume-law entanglement). Они впервые предложили использовать Ограниченную машину Больцмана (RBM) — простую двухслойную генеративную нейросеть — в качестве вариационного анзаца для волновой функции спиновых систем.
2. Прорывной момент (2016–2017 гг.)
Официальным рождением NNQS считается февраль 2017 года (хотя препринт появился в 2016). Физики Джузеппе Карлео (Giuseppe Carleo) и Маттиас Тройер (Matthias Troyer) опубликовали в журнале Science знаковую работу: «Solving the quantum many-body problem with artificial neural networks».
3. Эпоха глубоких сетей (2018–2022 гг.)
После успеха RBM исследователи быстро поняли, что мелкие (shallow) сети имеют свои ограничения, и начали внедрять архитектуры глубокого обучения (Deep Learning).
4. Трансформеры и современность (2023–2026 гг.)
В последние годы вектор развития сместился в сторону масштабных моделей и ab initio («из первых принципов») квантовой химии



В 2017 году два физика, Джузеппе Карло и Маттиас Тройер, предложили метод квантовых состояний нейронной сети (NNQS). Этот метод основан на использовании нейронных сетей для аппроксимации состояния квантовой системы. Уравнение Шредингера - хорошо известное уравнение, которое играет центральную роль в квантовой механике
\begin{eqnarray*}
i\hbar {\frac {\partial \psi }{\partial t}}={\hat {H}}\psi, 
\end{eqnarray*}
где $\psi$ - волновая функция, $\hat {H}$ - оператор Гамильтона, а $\hbar$ - приведенная постоянная Планка.

Давайте решим стационарное уравнение Шредингера, и после некоторых преобразований ожидаемое значение энергии будет равно
\begin{eqnarray*}
E = \frac{\langle \psi|\hat{H}|\psi\rangle}{\langle \psi|\psi \rangle}.
\end{eqnarray*}

Теперь давайте используем нейронную сеть в качестве волновой функции. Вместо использования функции потерь. Поскольку энергия не обязательно стремится к нулю, а скорее стремится к конечному минимальному значению, в некоторых работа энрегия не рассматривается как функция потерь, иногда называемая функцией затрат. Однако во многих работах она упоминается как функция потерь, чтобы подчеркнуть идею метода. Мы будем использовать энергию и обучать нейронную сеть, чтобы минимизировать энергетическое значение. Это достигается с помощью градиентного спуска, который регулирует веса нейронной сети

\begin{eqnarray*}
Loss = E = \frac{\langle \psi|\hat{H}|\psi\rangle}{\langle \psi|\psi \rangle} \rightarrow \text{min}.
\end{eqnarray*}

Ортогональные состояния в квантовой механике — это состояния, которые являются взаимно исключающими. Если квантовая система находится в одном из них, вероятность обнаружить её в другом равна нулю.В контексте нейросетевых квантовых состояний (NNQS) математическое условие ортогональности двух волновых функций $\vert{}\psi_A\rangle$ и $\vert{}\psi_B\rangle$ выражается через их скалярное произведение (интеграл перекрытия):$\langle \psi _{A}|{}\psi _{B}\rangle =\sum _{\sigma }\psi _{A}^{*}(\sigma )\psi _{B}(\sigma )=0$ где σ — конфигурации квантовых чисел (например, спинов), а ψ(σ) — комплексные амплитуды, которые вычисляет нейросеть.

Нейросеть легко находит основное (самое низкоэнергетическое) состояние системы. Чтобы найти первое возбужденное состояние, алгоритм должен искать минимум энергии в пространстве, строго ортогональном основному состоянию.

VMC (Вариационный Монте-Карло): Стандартный метод квантовой физики многих тел. В его основе лежит вариационный принцип: энергия любого пробного состояния Ψ всегда больше или равна энергии основного состояния E₀. VMC использует алгоритм Метрополиса (Монте-Карло) для сэмплирования квантовых конфигураций и вычисления многомерных интегралов (ожидаемых значений энергии).

Простыми словами: VMC — это алгоритм оптимизации, а NNQS — это математическая модель волновой функции, которую этот алгоритм оптимизирует.

# Масштабируемость штрафного коэффициента
Для поиска возбужденных состояний в функцию потерь (Loss) добавляют штраф: $E_{loss} = \langle\psi_B\vert{}H\vert{}\psi_B\rangle + \lambda \vert{}\langle\psi_A\vert{}\psi_B\rangle\vert{}^2$. Баланс $\lambda$: Если коэффициент $\lambda$ слишком мал, нейросеть свалится в основное состояние ($\psi _{A}$). Если слишком велик — оптимизатор будет минимизировать только перекрытие, игнорируя реальную энергию системы.

# Ортогональность по симметрии
Если основное и возбужденное состояния принадлежат разным неприводимым представлениям группы симметрии, они ортогональны автоматически.

# Водородоподобный атом

Для поиска радиальной части волновой функции атома водорода методом NNQS (Neural Network Quantum States) ключевым шагом является корректное представление энергии через полный трехмерный гамильтониан в сферических координатах $(r, \theta, \varphi)$:

$$\hat{H}=-\frac{\hbar ^{2}}{2\mu }\left[\frac{1}{r^{2}}\frac{\partial }{\partial r}\left(r^{2}\frac{\partial }{\partial r}\right)+\frac{1}{r^{2}\sin \theta }\frac{\partial }{\partial \theta }\left(\sin \theta \frac{\partial }{\partial \theta }\right)+\frac{1}{r^{2}\sin ^{2}\theta }\frac{\partial ^{2}}{\partial \varphi ^{2}}\right]-\frac{Ze^{2}}{r}$$

Здесь $\mu$ — приведенная масса электрона, а потенциал описывает кулоновское притяжение к ядру, $Z$ - атомный номер, $\hbar$ - приведенная постоянная Планка и $e$ - электрический заряд электрона. 

Поскольку угловая часть описывается сферическими гармониками $\hat{L}^2 Y_{lm} = \hbar^2 l(l+1) Y_{lm}$, трехмерная задача часто сводится к одномерному радиальному уравнению Шрёдингера для функции $R(r)$.

$$\hat{H}=-\frac{\hbar^2}{2\mu}\frac{1}{r^2}\frac{\partial}{\partial r}\bigg(r^2\frac{\partial}{\partial r}\bigg)+\frac{l(l+1)\hbar^2}{2\mu r^2}-\frac{Ze^2}{r}.$$

Полученный результат принципиально важен: он позволяет свести гамильтониан для конкретного значения $l$ к функции одной переменной $r$. Таким образом, задача становится строго одномерной. Далее для удобства расчетов мы введем новую функцию:

$$ P_{nl}(r)=R_{nl}{r}. $$


В рамках данного подхода нейросетевые квантовые состояния (NNQS) выступают аналогом радиальных функций $P_{nl}(r)$. Для оптимизации вычислений целесообразно перейти к атомной системе единиц, в которой $m_e = e = \hbar = 1$ (где $m_{e}$ — масса покоя электрона). С учетом приближения $\mu \approx m_e$ окончательное выражение для гамильтониана в атомных единицах принимает вид:

$$\hat{H}=-\frac{1}{2}\frac{d^2}{dr^2}+\frac{l(l+1)}{2r^2}-\frac{Z}{r}.$$

# Моделирование водородоподобных систем методом NNQS

Реализация метода нейросетевых квантовых состояний (Neural Network Quantum States) для численного решения уравнения Шрёдингера.

## 🔬 Архитектура и методы
В проекте реализован классический пайплайн вариационного квантового Монте-Карло (VMC):
1. Сэмплирование конфигураций электронного облака (алгоритм Метрополиса-Гастингса).
2. Расчет локальной энергии (вычисление кинетической энергии через автоматическое дифференцирование).
3. Оптимизация параметров нейросети.

**Ключевая особенность:** В репозитории используется оригинальная (авторская) архитектура нейронной сети и оптимизированная логика подбора параметров, адаптированная под особенности кулоновского потенциала водородоподобных атомов.


```python
# import
import numpy as np
from tqdm.notebook import tqdm 
import torch
from torch import nn
from torch.autograd import grad
import matplotlib.pyplot as plt 

import os    
os.environ['KMP_DUPLICATE_LIB_OK'] = 'True'


# device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
# seeds
def set_random_seed(seed):
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    np.random.seed(seed)
    torch.backends.cudnn.deterministic = True
```

```python
Z = 1 # Charge number
alpha = 7.297e-3 # Fine-structure constant



# Network size
input_size = 1   # Inputs to the neural network
output_size = 1  # Outputs of the neural network
hidden_size = 50 # Nodes in the hidden layer

# Define the seed 
seed = 42 
set_random_seed(seed)
# Determine the coefficients for NN 
W1 = torch.rand(hidden_size, input_size, requires_grad=True)        
B1 = torch.rand(hidden_size, requires_grad=True)  - 0.5    
W2 = torch.rand(output_size, hidden_size, requires_grad=True) 
B2 = torch.rand(output_size, requires_grad=True)  - 0.5 
```

```python
# Mesh parameters
n_points = 10000                   # Number of points
left = 0                  # Lower limit
right = 30                 # Upper limit
h = (right - left) / (n_points - 1)  # Grid step
wi = torch.empty(n_points, 1).fill_(h).to(device) # Integration weights for this mesh. Dirichlet boundary conditions 
X_train = torch.linspace(left+h, right, n_points, requires_grad=True, device=device)
X = X_train.clone().unsqueeze(1)
x_cpu = X_train.detach().cpu().numpy()
```

## Нейронные сети

Во всех случаях мы применяем многослойный перцептрон (MLP) с одним скрытым слоем (50–100 нейронов для разных задач). По нашему мнению, для наших задач достаточно одного скрытого слоя, хотя можно использовать и два. В качестве функции активации мы используем гиперболический тангенс (tanh). Существуют и другие функции активации, такие как ReLU, SiLU и другие. Выбор зависит от конкретной задачи. В нашем исследовании мы выбрали функцию активации с гиперболическим тангенсом из-за симметричной области определения. Кроме того, мы используем планировщик для уменьшения скорости обучения и L2-регуляризацию, чтобы избежать нестабильности, связанной с ростом весов.

```python
class NNQS(nn.Module):
    def __init__(self, input_size, output_size, hidden_size, W1, B1, W2, B2):
        super().__init__()
        self.activation = nn.Tanh()
        self.fc1 = nn.Linear(input_size, hidden_size)      
        self.fc2 = nn.Linear(hidden_size, output_size) 
           
        with torch.no_grad():
            self.fc1.weight.copy_(W1)
            self.fc1.bias.copy_(B1)
            self.fc2.weight.copy_(W2)
            self.fc2.bias.copy_(B2)
   
    # Set the output
    def forward(self, x):
        out1 = self.fc1(x)
        out = self.fc2(self.activation(out1))
        return out
```

Явный вид функций $P_{nl}(r)$ для первых уровней:

$$ P_{10}(r) = 2Z^{3/2}re^{-Zr}, $$

$$ P_{20}(r) = \frac{1}{\sqrt{2}}Z^{3/2}r\bigg(1-\frac{1}{2}Zr\bigg)e^{-Zr/2}, $$

$$ P_{21}(r) = \frac{1}{2\sqrt{6}}Z^{5/2}r^2e^{-Zr/2}. $$


```python
R_10 = lambda r : 2.*np.power(Z, 1.5)*r*np.exp(-Z*r)
R_20 = lambda r : (1/np.sqrt(2.))*np.power(Z, 1.5)*r*(1-0.5*Z*r)*np.exp(-Z*r/2.)
R_21 = lambda r : (1/(2*np.sqrt(6.)))*np.power(Z, 2.5)*np.power(r, 2)*np.exp(-Z*r/2.)
```

```python
E0_theory = -0.5*Z**2
E1_theory = -1./8.*Z**2
E2_theory = -1./8.*Z**2

print("E_0=", E0_theory)
print("E_1=", E1_theory)
print("E_2=", E2_theory)
```

## Функция потерь (Loss function)

Функция потерь вычисляется  

$$Loss=\frac{\langle \psi_{nl,\theta} |\hat{H}| \psi_{nl,\theta} \rangle}{\langle{\psi_{nl,\theta}}|\psi_{nl,\theta}\rangle}. $$

Как ранее было сказано, задачу можно свести к одномерной (интегрируем первое слагаемое по частям)

$$ \langle\psi_{nl,\theta}|\hat{H}|\psi_{nl,\theta}\rangle = \int_0^\infty \bigg\{\frac{1}{2}\bigg(\frac{d\psi_{nl,\theta}(r)}{dr}\bigg)^2+\psi_{nl,\theta}^2(r)\bigg[\frac{l(l+1)}{2r^2}-\frac{Z}{r}\bigg]\bigg\}dr $$

При обычной дискретизации они записываются:
$$ \langle\psi_{nl,\theta}|\hat{H}|\psi_{nl,\theta} \rangle \approx \sum_{i=1}^{N_r} w_i \bigg[\frac{1}{2}\bigg(\frac{d\psi_{nl,\theta}(r)}{dr}\bigg)^2\bigg|_{r=r_i}+\psi_{nl,\theta}^2(r_i)\bigg(\frac{l(l+1)}{2r_i^2}-\frac{Z}{r_i}\bigg)\bigg], $$

$$N = \langle \psi_{nl,\theta}|\psi_{nl,\theta}\rangle\approx \sum_{i=1}^{N_r} w_i \psi_{nl,\theta}^2(r_i).$$

Здесь через $w_{i}$ обозначены весовые коэффициенты интегрирования на расчетной сетке, переменная $N_{r}$ задает общее число дискретных точек, а величина $N$ выполняет роль нормировочного множителя.

В явном виде заложить в нейросетевую модель граничное условие $\psi_{nl}(0)=0$ позволяет использование специальной огибающей функции, умножаемой на выход NNQS. Этот подход полностью заимствован из классического аналитического формализма. Внедрение такого сомножителя трансформирует шаг оптимизации: теперь физические краевые ограничения удовлетворяются автоматически прямо в процессе минимизации лосс-функции, избавляя сеть от необходимости тратить вычислительные ресурсы на поиск правильного поведения функции в нуле.

$$ \psi_{nl}(r)= network(r)\cdot r^{l+1}e^{-Zr/n}  $$


```python
def loss_1(l): 
    lambd = 0
    psi = net(X)*X.pow(l+1)*torch.exp(-Z*X)
    psi2 = psi.pow(2)
    psi_r = grad(outputs=psi, inputs=X,grad_outputs=torch.ones_like(psi),create_graph=True)[0]
    K = 0.5*torch.tensordot(psi_r.pow(2), wi) 
    U = torch.tensordot(psi2 * ( l*(l+1)/(2*X.pow(2)) - Z/X), wi) 
    N = torch.tensordot(psi2, wi) 
    regul = torch.tensordot(net.fc1.weight,net.fc1.weight) 
    + torch.tensordot(net.fc2.weight,net.fc2.weight)  # regularization
    
    return (K + U)/N + lambd*regul, K/N, U/N, psi, psi / torch.sqrt(N)
```

## Training process

The $1s$ level n= 0 and l = 0

Для расчета первого возбужденного состояния $(n=2, l=0)$ используется аналогичная схема оптимизации, дополненная условием строгой ортогональности к основному состоянию:

$$ \langle\psi_{10,\theta}|\psi_{20,\theta'}\rangle = 0, $$

где $\psi_{10,\theta}$ и $\psi_{20,\theta'}$ — волновые функции основного и первого возбужденного состояний соответственно.  Реализовать данное ограничение можно за счет введения неопределенного множителя Лагранжа, который накладывает штраф за нарушение ортогональности прямо в процессе минимизации функционала потерь 

$$ Loss = \frac{\langle{\psi_{20,\theta'}}|\hat{H}|{\psi_{20,\theta'}}\rangle}{\langle\psi_{20,\theta'}|\psi_{20,\theta'}\rangle} + \lambda\frac{\langle\psi_{\theta,10}|\psi_{20,\theta'}\rangle^2}{\langle\psi_{20,\theta'}|\psi_{20,\theta'}\rangle\langle\psi_{10,\theta}|\psi_{10,\theta}\rangle} + L_{2,regul}. $$

 

```python
def loss_excited(l):  
    lambd = 0
    psi = net(X)*X.pow(l+1)*torch.exp(-Z*X/2)
    psi2 = psi.pow(2)
    psi_r = grad(outputs=psi, inputs=X,grad_outputs=torch.ones_like(psi),create_graph=True)[0]   
    K = 0.5*torch.tensordot(psi_r.pow(2), wi)  
    U = torch.tensordot(psi2 * ( l*(l+1)/(2*X.pow(2)) - Z/X ), wi) 
    N = torch.tensordot(psi2, wi) 
    mul1 = (torch.tensordot(psi*psi_10, wi)).pow(2)
    mul2 = (torch.tensordot(psi2, wi)) * (torch.tensordot(psi_10.pow(2), wi))
    ort = lambda_L * (mul1/mul2)
    regul = torch.tensordot(net.fc1.weight,net.fc1.weight) 
    + torch.tensordot(net.fc2.weight,net.fc2.weight) #regularization
    
    return (K + U) / N + ort +lambd * regul,  K / N, U / N, psi, psi / torch.sqrt(N)
```

# Релятивистская бесспиновая частица в потенциале

В релятивистской квантовой механике для свободной частицы с энергией массы покоя соотношение между энергией $E$ и импульсом $p$
\begin{equation}
E^2 = p^2 c^ 2 + m^ 2c^ 4,
\end{equation}
где $c$ — скорость света в вакууме, $m$ — масса частицы. Используя потенциал $V(x)$, мы получаем
\begin{equation}
[E-V]^2 = p^ 2 c ^ 2 + m ^ 2c ^ 4.
\end{equation}

Соответствующим волновым уравнением в одном измерении является
\begin{equation}
[E-V]^2 \psi(x) = \left[ -\hbar^2 c^2 \frac{d^2}{dx^2}+ m^ 2c^4 \right]\psi(x),
\end{equation}
где $\hbar$ — приведенная постоянная Планка, $\psi(x)$ — волновая функция. 

Это так называемое релятивистское уравнение Клейна—Гордона—Фока. Мы можем записать его в виде уравнения, подобного уравнению Шредингера 

\begin{equation}
E_{eff} \psi(x) = \left[ -\frac{\hbar^2}{2m}  \frac{d^2}{dx^2}+ V_{eff}(x) \right]\psi(x), 
\end{equation}
где мы обозначили 
\begin{equation}
E_{eff} = \frac{E^2-m^2c^4}{2mc^2},
\end{equation}

\begin{equation}
V_{eff} = \frac{2EV(x) - V^2(x)}{2mc^2},
\end{equation}

\begin{equation}
\hat{H}_{eff} = \left[ -\frac{\hbar^2}{2m}  \frac{d^2}{dx^2}+ V_{eff}(x) \right],
\end{equation}


\begin{equation}
E_{eff} = \frac{\langle \psi|\hat{H}_{eff}|\psi\rangle}{\langle \psi|\psi \rangle}\geq E_0,
\end{equation}
$E_0$ — минимальная энергия для нашей системы.
